## Data Cleaning

### Step 1: Missing Values Analysis

Before doing any imputation or removal, we first need to identify which columns 
contain missing (null) values and how significant the missingness is.

We calculate both the raw count and the percentage of missing values per column, 
sorted in descending order, to prioritize which columns need attention first.

```python
missing = football_df.isnull().sum()
missing_percent = (missing / len(football_df)) * 100

missing_df = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_percent.round(2)
}).sort_values("missing_percent", ascending=False)

missing_df[missing_df["missing_count"] > 0]
```

### Step 2: Categorizing Missing Values

Looking at the missingness percentages, the columns fall into three distinct 
categories, each requiring a different handling strategy:

**Category 1 — True zeros (attempt = 0 → rate is undefined)**

Many columns follow an "attempts + success_rate" pattern (e.g. `dribble_attempts` 
and `dribble_success_rate`). When a player made zero attempts, the success rate 
becomes NaN due to division by zero — this is not truly missing data, it simply 
means the player had zero occurrences of that action. These NaNs should be filled 
with 0.

**Category 2 — Structural missingness (data not collected for certain leagues)**

Columns like `expected_assists`, `xg_plus_xa`, and `shots_on_target` have very 
high missing percentages (53%–63%). To verify whether this is random or 
structural, we grouped the missing percentage by `league_name`. The result 
showed that many smaller/lower-tier leagues have 100% missing values for these 
columns, while bigger leagues (e.g. Ligue 1, Bundesliga) have much lower 
missingness (~10%–21%). This confirms that fotmob simply does not collect 
advanced stats (xG, xA, shots on target) for lower-tier leagues — it is not that 
players had zero of these stats.

**Category 3 — Critical/target-adjacent columns with low missingness**

Columns such as `minutes_played`, `pass_accuracy`, `team_goals`, `opponent_goals`, 
and `goal_difference` had a very small percentage of missing values (< 1.5%). 
Since filling these with 0 would be misleading (e.g. a player with missing 
`minutes_played` did not necessarily play 0 minutes), we chose to drop the rows 
with missing values in these columns instead, given the negligible data loss.

### Step 3: Filling Category 1 columns with 0

These columns represent counts or rates where NaN simply means "zero attempts", 
so we fill them with 0.

```python
zero_fill_cols = [
    "cross_attempts", "accurate_crosses", "cross_success_rate",
    "dribble_attempts", "successful_dribbles", "dribble_success_rate",
    "long_ball_attempts", "accurate_long_balls", "long_ball_success_rate",
    "aerial_duel_success_rate",
    "ground_duel_attempts", "ground_duels_won", "ground_duel_success_rate",
    "duels_won", "duels_lost",
    "touches_opp_box", "dribbled_past",
    "goals", "blocked_shots", "chances_created",
    "assists", "was_fouled",
    "aerial_duel_attempts", "aerial_duels_won"
]

football_df[zero_fill_cols] = football_df[zero_fill_cols].fillna(0)
```

### Step 4: Verifying Structural Missingness by League

To confirm our hypothesis for `expected_assists`, `xg_plus_xa`, and 
`shots_on_target`, we grouped the missing percentage by league. The results 
confirmed a clear league-dependent pattern (100% missing for lower-tier leagues, 
much lower for major leagues).

```python
missing_by_league = football_df.groupby("league_name")["expected_assists"].apply(
    lambda x: x.isnull().mean() * 100
).sort_values(ascending=False)

print(missing_by_league)
```

### Step 5: Final Decision on Category 2 Columns

- `expected_assists` and `xg_plus_xa`: dropped entirely. Both have extremely 
  high, league-dependent missingness (~59%–63%), and imputing them (with 0 or 
  mean) would introduce strong bias tied to league tier rather than actual 
  player performance.

- `passes_final_third`: dropped. This column was judged to add little 
  explanatory value for predicting `player_rating` and also carried notable 
  missingness (~15%).

- `shots_on_target`: kept, since shots on target is historically one of the 
  most important attacking indicators in football and is a player-level 
  (not team-level) statistic. Instead of dropping it, we created a binary flag 
  column `has_shot_data` to indicate whether this stat was actually available 
  for that match/league, then filled the remaining NaNs with 0. This preserves 
  the signal for leagues where the data exists, while letting the model 
  distinguish "zero shots" from "data not available" through the flag column.

```python
football_df = football_df.drop(columns=["expected_assists", "xg_plus_xa", "passes_final_third"])
football_df["has_shot_data"] = football_df["shots_on_target"].notnull().astype(int)
football_df["shots_on_target"] = football_df["shots_on_target"].fillna(0)
```

```python
missing = football_df.isnull().sum()
missing_percent = (missing / len(football_df)) * 100

missing_df = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_percent.round(2)
}).sort_values("missing_percent", ascending=False)

missing_df[missing_df["missing_count"] > 0]
```

### Step 6: Applying Final Column Decisions

```python
critical_cols = ["minutes_played", "pass_accuracy", "opponent_goals", "team_goals", "goal_difference"]
football_df = football_df.dropna(subset=critical_cols)
```

After applying all the above steps, we re-checked the total number of missing 
values in the dataset, confirming it dropped to 0. Missing values handling is 
now complete.

## ML model
### Step 11: Creating the ML-ready DataFrame

We create a separate DataFrame dedicated to machine learning, dropping all 
identifier and metadata columns (IDs, names, dates, league/country info) that 
have no predictive value for `player_rating`.

We build **two versions** to test whether team-level match outcome should be 
included as a feature:

- **Version A (`ml_df_no_result`)**: Contains only individual player performance 
  statistics (passing, dribbling, defensive actions, duels, fouls, etc.), 
  completely excluding any information about whether the team won or lost.

- **Version B (`ml_df_with_result`)**: Same as Version A, but additionally 
  includes `goal_difference`, which summarizes the match outcome (positive = 
  win, negative = loss, zero = draw).

We will train the model on both versions and compare their performance and 
feature importances, to determine whether including team result information 
meaningfully improves predictions or introduces unwanted bias (the model 
relying on team outcome rather than genuine individual performance).

Additionally, we removed goalkeepers (`is_goalkeeper == 1`) from both versions. 
Goalkeeper statistics carry a fundamentally different meaning than outfield 
player statistics (e.g. `pass_accuracy` reflects distribution style rather than 
playmaking quality), and the dataset lacks goalkeeper-specific metrics (saves, 
goals conceded, clean sheets) that would be necessary to properly evaluate 
their performance. Since goalkeepers make up only ~7.5% of the dataset, 
removing them has minimal impact on data volume while producing a cleaner, 
more consistent model for outfield players.

```python
# columns to always drop for ML (IDs and metadata with no predictive value)
id_cols_to_drop = [
    "match_id", "match_date",
    "league_id", "league_name", "country_code",
    "parent_league_name", "parent_league_season",
    "home_team_id", "home_team_name",
    "away_team_id", "away_team_name",
    "player_id", "player_name",
    "team_id", "team_name"
]

# redundant columns representing the same "team result" concept
redundant_result_cols = ["home_score", "away_score", "team_goals", "opponent_goals", "team_result"]

# ============ VERSION A: without team result info (player performance only) ============
ml_df_no_result = football_df.drop(columns=id_cols_to_drop + redundant_result_cols + ["goal_difference"])

# ============ VERSION B: with team result info (goal_difference included) ============
ml_df_with_result = football_df.drop(columns=id_cols_to_drop + ["home_score", "away_score", "team_goals", "opponent_goals", "team_result"])

# ============ Remove goalkeepers from both versions ============
ml_df_no_result = ml_df_no_result[ml_df_no_result["is_goalkeeper"] == 0].drop(columns=["is_goalkeeper"])
ml_df_with_result = ml_df_with_result[ml_df_with_result["is_goalkeeper"] == 0].drop(columns=["is_goalkeeper"])

print("Version A (no team result) shape:", ml_df_no_result.shape)
print(ml_df_no_result.columns.tolist())

print("\nVersion B (with goal_difference) shape:", ml_df_with_result.shape)
print(ml_df_with_result.columns.tolist())
```

### Step 12: Encoding Check and Train-Test Split

Before splitting the data, we verified that no categorical (object-type) columns 
remain in the ML-ready DataFrames, since all identifier and text columns 
(player name, team name, league name, etc.) were already removed in the previous 
step. The remaining columns are all numeric, including binary indicators 
(`is_home_team`, `has_shot_data`), so no additional encoding is required.

We then split both versions of the dataset (with and without team result 
information) into training and testing sets using an 80/20 split, with 
`player_rating` as the target variable (`y`) and all remaining columns as 
features (`X`). A fixed `random_state` was used to ensure reproducibility.

```python
from sklearn.model_selection import train_test_split

# ============ Version A: without team result ============
X_a = ml_df_no_result.drop(columns=["player_rating"])
y_a = ml_df_no_result["player_rating"]

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(
    X_a, y_a, test_size=0.2, random_state=42
)

# ============ Version B: with team result (goal_difference) ============
X_b = ml_df_with_result.drop(columns=["player_rating"])
y_b = ml_df_with_result["player_rating"]

X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_b, y_b, test_size=0.2, random_state=42
)

print("Version A -> train:", X_train_a.shape, "test:", X_test_a.shape)
print("Version B -> train:", X_train_b.shape, "test:", X_test_b.shape)
```

### Step 13: Feature Scaling

We apply `StandardScaler` to standardize all numeric features (mean = 0, 
std = 1). This is important because features in this dataset have very 
different scales (e.g. `touches` ranges up to ~199, while `pass_accuracy` 
ranges 0-100, and `goals` ranges 0-4), which can bias distance-based or 
linear models toward features with larger numeric ranges.

Note: the scaler is fit only on the training data (`fit_transform`) and then 
applied to the test data using `transform` only (without re-fitting). This 
prevents data leakage, since fitting the scaler on test data would let 
information about the test set's distribution influence the training process.

```python
from sklearn.preprocessing import StandardScaler

# ============ Version A ============
scaler_a = StandardScaler()
X_train_a_scaled = scaler_a.fit_transform(X_train_a)
X_test_a_scaled = scaler_a.transform(X_test_a)

# ============ Version B ============
scaler_b = StandardScaler()
X_train_b_scaled = scaler_b.fit_transform(X_train_b)
X_test_b_scaled = scaler_b.transform(X_test_b)

print("Version A scaled shape:", X_train_a_scaled.shape)
print("Version B scaled shape:", X_train_b_scaled.shape)
```

### Step 14: Model Training with XGBoost

We train an `XGBRegressor` model separately on both dataset versions (with and 
without team result information) to predict `player_rating`. Since XGBoost is 
a tree-based model, feature scaling is not required — the raw (unscaled) 
features are used directly, which also keeps feature importance values more 
interpretable later.

We evaluate each model using three metrics:
- **MAE** (Mean Absolute Error): average absolute difference between predicted 
  and actual ratings.
- **RMSE** (Root Mean Squared Error): similar to MAE but penalizes larger 
  errors more heavily.
- **R²** (R-squared): proportion of variance in `player_rating` explained by 
  the model (closer to 1 is better).

Comparing these metrics between Version A and Version B will help us determine 
whether including team-level match outcome (`goal_difference`) meaningfully 
improves prediction accuracy, or whether it mainly reflects data leakage.

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# ============ Version A: without team result ============
model_a = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    n_jobs=-1
)

model_a.fit(X_train_a, y_train_a)
y_pred_a = model_a.predict(X_test_a)

mae_a = mean_absolute_error(y_test_a, y_pred_a)
rmse_a = np.sqrt(mean_squared_error(y_test_a, y_pred_a))
r2_a = r2_score(y_test_a, y_pred_a)

print("Version A (no team result):")
print("MAE:", round(mae_a, 4))
print("RMSE:", round(rmse_a, 4))
print("R2:", round(r2_a, 4))
```

```python
# ============ Version B: with team result (goal_difference) ============
model_b = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    n_jobs=-1
)

model_b.fit(X_train_b, y_train_b)
y_pred_b = model_b.predict(X_test_b)

mae_b = mean_absolute_error(y_test_b, y_pred_b)
rmse_b = np.sqrt(mean_squared_error(y_test_b, y_pred_b))
r2_b = r2_score(y_test_b, y_pred_b)

print("Version B (with goal_difference):")
print("MAE:", round(mae_b, 4))
print("RMSE:", round(rmse_b, 4))
print("R2:", round(r2_b, 4))
```

```python
import pandas as pd

# Feature importance for Version B
importance_b = pd.DataFrame({
    "feature": X_b.columns,
    "importance": model_b.feature_importances_
}).sort_values("importance", ascending=False)

print(importance_b.head(15))
```

### Step 15: Comparing Version A vs Version B — Conclusion

**Results Summary**

| Metric | Version A (no team result) | Version B (with goal_difference) |
|--------|------------------------------|-----------------------------------|
| MAE    | 0.2396                       | 0.1533                            |
| RMSE   | 0.3617                       | 0.2427                            |
| R²     | 0.7535                       | 0.8891                            |

Version B shows a notably better performance across all three metrics, with R² 
improving from ~75% to ~89% after adding `goal_difference` as a feature.

**Feature Importance Check**

To investigate whether this improvement stems from genuine signal or from data 
leakage, we examined the feature importances of Version B. `goal_difference` 
ranked 6th overall with an importance of ~0.053 — meaningful, but far from 
dominant. The top features (`goals`, `assists`, `touches`, `duels_won`, 
`chances_created`) are all legitimate individual performance statistics and 
together account for the majority of the model's decision-making.

**Conclusion**

Since `goal_difference` does not overwhelmingly dominate feature importance 
(unlike what we would expect from severe leakage), we conclude that its 
contribution reflects a real, moderate effect: match outcome does provide 
some additional context for player rating (which aligns with real-world 
football scoring conventions, where players on winning teams tend to receive 
slightly higher ratings even with similar individual stats). However, since 
the project's primary goal is to evaluate **individual player performance**, 
we consider Version A (without team result) to be the more appropriate and 
interpretable model for this purpose, as it isolates the player's own 
contribution from the team's overall match outcome.

We will proceed with **Version A** as our primary model for further evaluation 
and interpretation, while keeping Version B as a secondary comparison to show 
the effect of match context in our final report/notebook.

### Step 16: Hyperparameter Tuning with GridSearchCV

To improve the baseline XGBoost model (Version A), we performed hyperparameter 
tuning using `GridSearchCV` with 3-fold cross-validation, searching over key 
parameters: `n_estimators`, `max_depth`, `learning_rate`, and `subsample`. We 
used `neg_mean_absolute_error` as the scoring metric since minimizing prediction 
error is our primary objective.

After identifying the best parameter combination, we re-evaluated the tuned 
model on the held-out test set to compare it against the baseline model's 
performance (MAE: 0.2396, RMSE: 0.3617, R²: 0.7535).

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [200, 300, 500],
    "max_depth": [4, 6, 8],
    "learning_rate": [0.01, 0.05, 0.1],
    "subsample": [0.8, 1.0]
}

xgb_base = XGBRegressor(random_state=42, n_jobs=-1)

grid_search = GridSearchCV(
    estimator=xgb_base,
    param_grid=param_grid,
    scoring="neg_mean_absolute_error",
    cv=3,
    verbose=2,
    n_jobs=-1
)

grid_search.fit(X_train_a, y_train_a)

print("Best parameters:", grid_search.best_params_)
print("Best CV MAE:", -grid_search.best_score_)
```

```python
best_model_a = grid_search.best_estimator_
y_pred_best_a = best_model_a.predict(X_test_a)

mae_best = mean_absolute_error(y_test_a, y_pred_best_a)
rmse_best = np.sqrt(mean_squared_error(y_test_a, y_pred_best_a))
r2_best = r2_score(y_test_a, y_pred_best_a)

print("Tuned Model (Version A):")
print("MAE:", round(mae_best, 4))
print("RMSE:", round(rmse_best, 4))
print("R2:", round(r2_best, 4))
```

### Step 17: Tuning Results and Conclusion

**Comparison: Baseline vs Tuned Model (Version A)**

| Metric | Baseline | Tuned (GridSearchCV) |
|--------|----------|------------------------|
| MAE    | 0.2396   | 0.2377                 |
| RMSE   | 0.3617   | 0.3600                 |
| R²     | 0.7535   | 0.7559                 |

The best parameters found were: `learning_rate=0.1`, `max_depth=4`, 
`n_estimators=500`, `subsample=1.0`.

The improvement from tuning was marginal (less than 0.5% across all metrics). 
This suggests that our initial baseline configuration was already close to 
optimal for this dataset. The tuned model favors a shallower tree depth (4 vs 
6) combined with more estimators (500 vs 300), which slightly reduces the risk 
of overfitting while maintaining similar predictive power. The small gap 
between baseline and tuned performance also indicates the model is relatively 
robust and not overly sensitive to hyperparameter choices.

We proceed with the tuned model (`best_model_a`) as our final model for 
Version A.

### Step 22: Second Model — Random Forest (Version A)

To validate and compare against our primary XGBoost model, we trained a second 
model using `RandomForestRegressor` on Version A (individual player performance 
features only, without team result).

Unlike XGBoost (Gradient Boosting, where trees are built sequentially to 
correct previous errors), Random Forest builds many independent trees in 
parallel and averages their predictions (Bagging). This makes it naturally 
more resistant to overfitting and less sensitive to hyperparameter tuning, 
though it can sometimes be less precise than a well-tuned boosting model.

We will compare its performance (MAE, RMSE, R²) against the tuned XGBoost 
model to check whether the results are consistent across different model 
architectures — which would strengthen confidence in our feature selection 
and overall approach.

```python
from sklearn.ensemble import RandomForestRegressor

rf_model_a = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    random_state=42,
    n_jobs=-1
)

rf_model_a.fit(X_train_a, y_train_a)
y_pred_rf_a = rf_model_a.predict(X_test_a)

mae_rf_a = mean_absolute_error(y_test_a, y_pred_rf_a)
rmse_rf_a = np.sqrt(mean_squared_error(y_test_a, y_pred_rf_a))
r2_rf_a = r2_score(y_test_a, y_pred_rf_a)

print("Random Forest (Version A):")
print("MAE:", round(mae_rf_a, 4))
print("RMSE:", round(rmse_rf_a, 4))
print("R2:", round(r2_rf_a, 4))
```
